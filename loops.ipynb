```python
def ex1():
    count = int(input("Enter number of times: "))
    char = input("Enter character: ")
    print(char * count)

def ex2():
    a = int(input("Enter range start (a): "))
    b = int(input("Enter range end (b): "))
    for i in range(a, b + 1):
        print(i, end=" ")
    for i in range(b - 1, a - 1, -1):
        print(i, end=" ")
    print()

def ex3_4_5():
    n = int(input("Enter n for list operations: "))
    
    list_a = list(range(1, n + 1))
    print("3a List:", list_a)
    
    list_b = [x**2 for x in list_a]
    print("3b Squares:", list_b)
    
    list_c = [x**2 if x % 2 != 0 else x // 2 for x in list_a]
    print("3c Odds squared / Evens halved:", list_c)
    
    filtered_list = [val for idx, val in enumerate(list_a) if (idx + 1) % 5 != 0]
    print("4 Without every 5th element:", filtered_list)
    
    m3 = sum(1 for x in list_a if x % 3 == 0)
    m5 = sum(1 for x in list_a if x % 5 == 0)
    mb = sum(1 for x in list_a if x % 3 == 0 and x % 5 == 0)
    me = sum(1 for x in list_a if x % 3 == 0 or x % 5 == 0)
    print("5 Multiples of 3:", m3, "| Multiples of 5:", m5, "| Both:", mb, "| Either:", me)

def ex6():
    print("Enter 10 numbers:")
    total = sum(float(input(f"Number {i + 1}: ")) for i in range(10))
    print("Sum:", total)

def ex7():
    print("First 10 powers of 2:", [2**i for i in range(10)])

def ex8():
    n = int(input("Enter number for factorial: "))
    fact = 1
    for i in range(1, n + 1):
        fact *= i
    print(f"{n}! =", fact)

def ex9():
    n = int(input("Enter n for Fib(n): "))
    if n == 0 or n == 1:
        print(f"F_{n} =", 1)
    else:
        a, b = 1, 1
        for _ in range(2, n + 1):
            a, b = b, a + b
        print(f"F_{n} =", b)

if __name__ == "__main__":
    ex1()
    ex2()
    ex3_4_5()
    ex6()
    ex7()
    ex8()
    ex9()
```
